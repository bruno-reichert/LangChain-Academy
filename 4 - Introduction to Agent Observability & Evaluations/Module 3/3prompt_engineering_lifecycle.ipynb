### Update our Application to use Prompt Hub

We're going to pretty much define the same RAG application as before - with one crucial improvement.

Instead of pulling our `RAG_PROMPT` from utils.py, we're going to connect to the Prompt Hub in LangSmith.

Let's add the code snippet that will pull down our prompt that we just iterated on!

````python
import os
import json
import time
from pathlib import Path
from typing import List
import nest_asyncio
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

# Guardrail: prevents convert_prompt_to_openai_format from crashing on missing OpenAI key
os.environ.setdefault("OPENAI_API_KEY", "dummy_key_for_formatter")

from openai import OpenAI
from langsmith import traceable, Client
from langsmith.client import convert_prompt_to_openai_format
from langchain_core.documents import Document
from langchain_community.retrievers import TFIDFRetriever

# --- Stack Configuration ---
MODEL_NAME = "openai/gpt-oss-120b"
MODEL_PROVIDER = "groq"
APP_VERSION = 2.0  # Bumped for Prompt Hub versioning

openai_client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ.get("GROQ_API_KEY")
)

# --- Pull Prompt from LangSmith Prompt Hub ---
client = Client()

# TODO: If you saved your prompt under a different name in the Hub, put that name here!
PROMPT_NAME = "rag_with_code"  # e.g., "rag-prompt", "langsmith-rag", etc.

try:
    prompt = client.pull_prompt(PROMPT_NAME)
    print(f"✅ Successfully pulled '{PROMPT_NAME}' from Prompt Hub!")
except Exception as e:
    print(f"⚠️ Could not pull '{PROMPT_NAME}', falling back to local template: {e}")
    from langchain_core.prompts import ChatPromptTemplate
    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are an assistant for question-answering tasks. Use the retrieved context to answer concisely. Whenever possible, provide a Python code example to help the user get started."),
        ("human", "Context: {context}\n\nQuestion: {question}")
    ])


# --- Sub-second Parquet Retriever (Rule #8) ---
def get_vector_db_retriever():
    candidates = [
        Path("resources/union.parquet"),
        Path("./union.parquet"),
        Path("../resources/union.parquet"),
        Path(__file__).resolve().parent / "resources" / "union.parquet" if "__file__" in locals() else None,
        Path.cwd() / "resources" / "union.parquet",
    ]
    target_path = next((c for c in candidates if c and c.exists()), None)
    if not target_path:
        found = list(Path.cwd().rglob("union.parquet"))
        if found:
            target_path = found[0]

    df = pd.read_parquet(target_path)
    text_col = "texts" if "texts" in df.columns else "text"
    meta_col = "metadatas" if "metadatas" in df.columns else "metadata"

    raw_texts = df[text_col].tolist()
    raw_metas = df[meta_col].tolist() if meta_col in df.columns else [{}] * len(raw_texts)

    docs = []
    for text, meta in zip(raw_texts, raw_metas):
        if isinstance(meta, str):
            try:
                meta = json.loads(meta)
            except Exception:
                meta = {}
        docs.append(Document(page_content=str(text), metadata=meta if isinstance(meta, dict) else {}))

    return TFIDFRetriever.from_documents(docs, k=4)

nest_asyncio.apply()
retriever = get_vector_db_retriever()


# --- Tracing Hierarchy ---

@traceable(run_type="chain")
def retrieve_documents(question: str):
    return retriever.invoke(question)


@traceable(run_type="chain")
def generate_response(question: str, documents):
    # Rule #5: Token Diet
    formatted_docs = "\n\n".join(doc.page_content[:600] for doc in documents)
    
    # Hydrate prompt pulled from Prompt Hub
    formatted_prompt = prompt.invoke({"context": formatted_docs, "question": question})
    messages = convert_prompt_to_openai_format(formatted_prompt)["messages"]
    
    return call_openai(messages)


@traceable(
    run_type="llm",
    metadata={
        "ls_provider": MODEL_PROVIDER,
        "ls_model_name": MODEL_NAME,
        "app_version": APP_VERSION,
    }
)
def call_openai(messages: List[dict]) -> str:
    return openai_client.chat.completions.create(
        model=MODEL_NAME,
        messages=messages,
        max_tokens=400,  # Keeps generation safe and unclipped
    )


@traceable(run_type="chain")
def langsmith_rag(question: str):
    documents = retrieve_documents(question)
    response = generate_response(question, documents)
    return response.choices[0].message.content


# Quick test invocation
test_q = "How do I set up tracing to LangSmith with @traceable?"
print("\n[Question]:", test_q)
print("[Answer]:\n", langsmith_rag(test_q))
````
