```python
from dotenv import load_dotenv, find_dotenv
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langchain.messages import HumanMessage, AIMessage
from pprint import pprint

load_dotenv(find_dotenv())
```

```python
# 1. Groq Model
model = init_chat_model(model='openai/gpt-oss-120b', model_provider='groq')
response = model.invoke('Hello! Hello! You there?')
print(response.content)
```

```python
# 2. Gemini Model & Agent
gemini = init_chat_model(model='models/gemini-3.5-flash', model_provider='google_genai')
agent = create_agent(model=gemini)

response = agent.invoke({
    'messages': [
        HumanMessage(content='What\'s a good place to buy a stapler? Mine is broken!'),
        AIMessage(content='Try a Staples! They have lots of good stuff.'),
        HumanMessage(content='But there isn\'t a Staples nearby! I live in Sheboygan! I\'m doomed! What do I do?')
    ]
})

ans = response['messages'][-1].content
print(ans[0]['text'] if isinstance(ans, list) else ans)
```

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me about Mars!")]},
    stream_mode="messages"
):
    if token.content:
        # If Gemini wraps it in a list, grab the text; otherwise print directly (Groq/OpenAI)
        if isinstance(token.content, list):
            text = token.content[0].get('text', '') if isinstance(token.content[0], dict) else str(token.content[0])
            print(text, end="", flush=True)
        else:
            print(token.content, end="", flush=True)
```

```python
# 3. System Prompts
system_prompt = "You are a science fiction writer, create a capital city at the users request."

gemini = init_chat_model(model='models/gemini-3.5-flash', model_provider='google_genai')
scifi_agent = create_agent(model=gemini, system_prompt=system_prompt)

question = HumanMessage(content="Create a capital city for the moon, please!")

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][-1].content[0]['text'])
```

```python
# 4. Few-shot Prompting
system_prompt = """

You are a science fiction writer, create a space capital city at the users request.

User: What is the capital of mars?
Scifi Writer: Marsialis

User: What is the capital of Venus?
Scifi Writer: Venusovia

"""

gemini = init_chat_model(model='models/gemini-3.5-flash', model_provider='google_genai')
scifi_agent = create_agent(model=gemini, system_prompt=system_prompt)

question = HumanMessage(content="Create a capital city for the moon, please!")

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][-1].content[0]['text'])
```

```python
system_prompt = """

You are a science fiction writer, create a space capital city at the users request.

Please keep to the below structure.

Name: The name of the capital city

Location: Where it is based

Vibe: 2-3 words to describe its vibe

Economy: Main industries

"""


gemini = init_chat_model(model='models/gemini-3.5-flash', model_provider='google_genai')
scifi_agent = create_agent(model=gemini, system_prompt=system_prompt)

question = HumanMessage(content="Create a capital city for a lush jungle planet, please!")

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][-1].content[0]['text'])
```

```python
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

gemini = init_chat_model(model='models/gemini-3.5-flash', model_provider='google_genai')
scifi_agent = create_agent(model=gemini, system_prompt=system_prompt, response_format=CapitalInfo)

question = HumanMessage(content="Create a capital city for a dry desert planet, please!")

response = scifi_agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
```

```python
response["structured_response"].name
```

```python
capital_info = response["structured_response"]

capital_name = capital_info.name
capital_location = capital_info.location

print(f"{capital_name} is a city located at {capital_location}")
```
